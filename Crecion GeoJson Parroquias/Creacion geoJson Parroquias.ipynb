```python
import geopandas as gpd
import unicodedata
import pandas as pd

gdf = gpd.read_file("ecu_adm_2024/ecu_adm_adm3_2024.shp")

print(len(gdf))
#fuente: https://data.humdata.org/dataset/cod-ab-ecu
```

```python
gdf.columns
```

```python
gdf = gdf[
    [
        "ADM3_ES",     # Parroquia
        "ADM2_ES",     # Cantón
        "ADM1_ES",     # Provincia
        "geometry"
    ]
]
```

```python
gdf.head()
```

```python
gdf["geometry"] = gdf["geometry"].simplify(0.01)
```

# Normalizando

```python
def normalizar(texto):

    if pd.isna(texto):
        return texto

    #Eliminar espacios al inicio y al final, y coloca en mayusculas
    texto = str(texto).strip().upper()

    # Eliminar tildes y diéresis
    texto = ''.join(
        c for c in unicodedata.normalize('NFD', texto)
        if unicodedata.category(c) != 'Mn'
    )

    # Eliminar espacios dobles
    texto = ' '.join(texto.split())

    return texto

gdf["ADM3_ES"] = gdf ["ADM3_ES"].apply(normalizar)
gdf ["ADM2_ES"] = gdf ["ADM2_ES"].apply(normalizar)
gdf ["ADM1_ES"] = gdf ["ADM1_ES"].apply(normalizar)
```

```python
gdf.to_file("parroquias_ecuador.geojson", driver="GeoJSON")
```
