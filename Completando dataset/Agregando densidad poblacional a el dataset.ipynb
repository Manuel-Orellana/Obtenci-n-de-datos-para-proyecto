```python
import pandas as pd
```

```python
df_datos= pd.read_csv("dataset_incompleto.csv")
df_datos = df_datos.drop(columns= ["Unnamed: 0"])
df_datos.head(3)
```

```python
df_datos.shape
```

```python
df_densidad = pd.read_csv("densidad_poblacional.csv", dtype= {"DPA_PARROQ": str})
df_densidad = df_densidad.dropna()
# df_densidad = df_densidad.drop(columns=["fid"])
df_densidad["DPA_PARROQ"] = df_densidad["DPA_PARROQ"].astype(str)

df_densidad.head()
```

```python
df_densidad.shape
```

```python
df_final = df_datos.merge(
    df_densidad[
        ["DPA_PARROQ", "Densidad Poblacional"]
    ],
    on=["DPA_PARROQ"],
    how="left"
)
df_final.head(3)
```

```python
df_final.shape
```

```python
df_final.to_csv("dataset.csv")
```
